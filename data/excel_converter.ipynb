```python
import re

def create_slug(string):
    # Convert to lowercase
    string = string.lower()
    # Replace spaces with hyphens
    string = string.replace(" ", "-")
    # Remove any characters that are not alphanumeric or hyphens
    string = re.sub(r'[^a-z0-9\-]', '', string)
    if string.startswith("_"):
        string = string[1:]
    return string

def capitalize_first_letter(string):
    return string[0].upper() + string[1:].lower()

def create_id(category_name):
    category_name = category_name.lower().replace(" ", "_")
    if category_name.startswith("_"):
        category_name = category_name[1:]
    if category_name.endswith("_"):
        category_name = category_name[1:]
    return category_name


categories = []
category_map = {}  # To map category names to IDs

for row in categories_unique:
    if type(row) == str:
        category_name = capitalize_first_letter(row)
        
        # Check if the category already exists in the category_map
        if category_name in category_map:
            continue  # Skip this category if it already exists

        category_handle = create_slug(row)
        
        # Generate the category_id by converting the name to lowercase and replacing spaces with underscores
        category_id = create_id(category_name)
        
        category_map[category_name] = category_id

        category = {
            "id": category_id,
            "name": category_name,
            "rank": 0,
            "category_children": [],
            "handle": category_handle
        }
        categories.append(category)
```

```python
def create_options(description):
    # Use a regular expression to split the description by "y", "/", or "-"
    values = re.split(r'\s*[y/-]\s*', str(description))
    
    # Create the options object
    options = [
        {
            "title": "Formato",
            "values": values
        }
    ]
    return options

def create_variants(options, base_price, currency="crc", inventory_quantity=100):
    variants = []
    # print(options)
    # Iterate over the values in the "Formato" option
    for value in options[0]["values"]:
        title = value  # Title is simply the value in this case
        variant = {
            "title": title,
            "prices": [
                {
                    "currency_code": currency,
                    "amount": base_price
                }
            ],
            "options": [
                {"value": value}
            ],
            "inventory_quantity": inventory_quantity,
            "manage_inventory": False
        }
        variants.append(variant)
    
    return variants
```

```python
import numpy as np
import pandas as pd

def is_nan(string):
    if pd.isna(string):
        return ""
    else:
        return string

def get_description(row):
    description = row["Detalles"]
    return is_nan(description)

def transform_tags(tag_string):
    tags = tag_string.split(',')
    return [{"value": tag.strip()} for tag in tags]

def find_product_by_handle(products, handle):
    for product in products:
        if product["handle"] == handle:
            return product
    return None

products = []
for _, row in catalog_sheet.iterrows():
    if pd.notna(row["Nombre"]):
        options = create_options(row["Descripción"])
        handle = create_slug(str(row["Nombre"]))
        product = {
                "title": row["Nombre"],
                "categories":[],
                "description": get_description(row),
                "handle": handle,
                "is_giftcard": False,
                "weight": None,  # Populate as necessary
                "images": [],  # Populate as necessary from the Excel
                "options": options,
                "variants": create_variants(options, 1950),
                "metadata": {}
            }

        # Assign category
        category_name = row["Categoria"]
        if category_name in category_map:
            product["categories"] = [{"id": category_map[category_name]}]

        # Handle labels/tags
        if pd.notna(row["Etiquetas"]):
            product["tags"] = transform_tags(row["Etiquetas"])
        else:
            product["tags"] = []

        # Step 4: Handle metadata
        metadata_fields = ["Ingredientes", "Indicaciones", "Cloud", "ATC", "Principio Activo", 
                        "% Impuesto", "Código Casa", "Nombre Casa", "Estupefacientes y Psicotrópicos"]
        for field in metadata_fields:
            if field in row and pd.notna(row[field]):
                product["metadata"][field] = row[field]

        # Check if product with the same handle exists
        existing_product = find_product_by_handle(products, handle)
        if existing_product:
            # Compare the number of tags and keep the one with more tags
            if len(product["tags"]) > len(existing_product["tags"]):
                products.remove(existing_product)
                products.append(product)
        else:
            products.append(product)
```

```python
import json
import numpy as np
import pandas as pd

def is_nan(string):
    if pd.isna(string):
        return ""
    else:
        return string

def get_description(row):
    description = row["Detalles"]
    return is_nan(description)

def transform_tags(tag_string):
    tags = tag_string.split(',')
    return [{"value": tag.strip()} for tag in tags]

def find_product_by_handle(products, handle):
    for product in products:
        if product["handle"] == handle:
            return product
    return None

def save_products_from_i_to_n(catalog_sheet, start, end, header, categories):
    products = []
    counter = 0

    for index, row in catalog_sheet.iterrows():
        if counter >= end - start + 1:
            break

        if index >= start and pd.notna(row["Nombre"]):
            options = create_options(row["Descripción"])
            handle = create_slug(str(row["Nombre"]))
            product = {
                "title": row["Nombre"],
                "categories":[],
                "description": get_description(row),
                "handle": handle,
                "is_giftcard": False,
                "weight": None,  # Populate as necessary
                "images": [],  # Populate as necessary from the Excel
                "options": options,
                "variants": create_variants(options, 1950),
                "metadata": {}
            }

             # Assign category
            category_name = row["Categoria"]
            if category_name in category_map:
                print("Estoy en categoria")
                new_categories = [{"id": category_map[category_name]}]
                print(new_categories)
                product["categories"] = product["categories"] + new_categories

            # Handle labels/tags
            if pd.notna(row["Etiquetas"]):
                product["tags"] = transform_tags(row["Etiquetas"])
            else:
                product["tags"] = []

            # Step 4: Handle metadata
            metadata_fields = ["Ingredientes", "Indicaciones", "Cloud", "ATC", "Principio Activo", 
                            "% Impuesto", "Código Casa", "Nombre Casa", "Estupefacientes y Psicotrópicos"]
            for field in metadata_fields:
                if field in row and pd.notna(row[field]):
                    product["metadata"][field] = row[field]

            # Check if product with the same handle exists
            existing_product = find_product_by_handle(products, handle)
            if existing_product:
                # Compare the number of tags and keep the one with more tags
                if len(product["tags"]) > len(existing_product["tags"]):
                    products.remove(existing_product)
                    products.append(product)
            else:
                products.append(product)
                counter += 1  # Increment the counter only if a new product is added

    # Step 5: Construct the final JSON structure
    seed_data = {**header, "products": products, "categories": categories}

    # Step 6: Create a dynamic filename based on the range
    filename = f'./seed_output_{start}_to_{end}.json'


    # Step 7: Save the JSON to a file
    with open(filename, 'w') as f:
        json.dump(seed_data, f, indent=2)

    print(f"JSON file '{filename}' created successfully!")

```
